```python
import pandas as pd

data = {
    "player": ["A", "B", "C", "D", "E"],
    "games": [20, 18, 22, 15, 12],
    "points": [410, 124, 330, 300, 190],
}
df = pd.DataFrame(data)
df["points_per_game"] = df["points"] / df["games"]
df.sort_values("points_per_game", ascending=False)
```

```python
import matplotlib.pyplot as plt

df.plot(x="player", y="points_per_game", kind="bar", legend=False)
df = df.sort_values("points_per_game", ascending=False)
plt.title("Points per game by player (practice data)")
plt.ylabel("Points per game")
plt.show()
```
